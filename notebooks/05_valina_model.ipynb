#### **BIBLIOTECAS**

```python
import pandas as pd 
import numpy as np 

import seaborn as sns 
import matplotlib.pyplot as plt 

from sklearn.metrics import roc_auc_score, roc_curve, accuracy_score, recall_score

from catboost import CatBoostClassifier, Pool


pd.set_option('display.max_columns', None)

SEED = 42
```

#### **CARREGANDO DADOS**

```python
ORIGEM_DADOS = '../data/gold'
```

```python
#  DADOS DE TREINAMENTO
X_train = pd.read_parquet(f'{ORIGEM_DADOS}/train.parquet').drop(columns=['Churn'])
y_train= pd.read_parquet(f'{ORIGEM_DADOS}/train.parquet')['Churn']

# DADOS DE VALIDACAO
X_val = pd.read_parquet(f'{ORIGEM_DADOS}/val.parquet').drop(columns=['Churn'])
y_val= pd.read_parquet(f'{ORIGEM_DADOS}/val.parquet')['Churn']

# DADOS DE TESTE
X_test = pd.read_parquet(f'{ORIGEM_DADOS}/test.parquet').drop(columns=['Churn'])
y_test= pd.read_parquet(f'{ORIGEM_DADOS}/test.parquet')['Churn']
```

#### **MODELO**

```python

CATEGORICAS = X_train.select_dtypes(include=['object', 'string']).columns.to_list()


pool_train = Pool(X_train, y_train, cat_features=CATEGORICAS)
pool_val = Pool(X_val, y_val, cat_features=CATEGORICAS)
pool_test = Pool(X_test, y_test, cat_features=CATEGORICAS)


# VAnila MODEL 
model = CatBoostClassifier(random_seed=SEED, verbose=200)


model.fit(pool_test, eval_set=pool_val, plot=True)

```
